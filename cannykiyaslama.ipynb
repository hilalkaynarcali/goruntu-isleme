```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
img = cv2.imread("image4.jpeg", 0)
blur = cv2.GaussianBlur(img, (5, 5), 1)
```

```python
t1 = time.time()
canny = cv2.Canny(blur, 100, 200)
t2 = time.time()
canny_sure = t2 - t1
```

```python
prewitt_x = np.array([[-1, 0, 1],
                      [-1, 0, 1],
                      [-1, 0, 1]])

sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

laplacian = np.array([[0,  1, 0],
                      [1, -4, 1],
                      [0,  1, 0]])

t1 = time.time()
prewitt = np.abs(cv2.filter2D(blur, cv2.CV_64F, prewitt_x))
t2 = time.time()
prewitt_sure = t2 - t1

t1 = time.time()
sobel = np.abs(cv2.filter2D(blur, cv2.CV_64F, sobel_x))
t2 = time.time()
sobel_sure = t2 - t1

t1 = time.time()
lap = np.abs(cv2.filter2D(blur, cv2.CV_64F, laplacian))
t2 = time.time()
lap_sure = t2 - t1
```

```python
log = np.array([[0,  0, -1, -1, -1,  0,  0],
                [0, -1, -3, -3, -3, -1,  0],
                [-1, -3,  0,  7,  0, -3, -1],
                [-1, -3,  7, 24,  7, -3, -1],
                [-1, -3,  0,  7,  0, -3, -1],
                [0, -1, -3, -3, -3, -1,  0],
                [0,  0, -1, -1, -1,  0,  0]])

t1 = time.time()
log_sonuc = np.abs(cv2.filter2D(img, cv2.CV_64F, log))
t2 = time.time()
log_sure = t2 - t1
```

```python
print("Prewitt :", prewitt_sure)
print("Sobel   :", sobel_sure)
print("Laplacian:", lap_sure)
print("LoG     :", log_sure)
print("Canny   :", canny_sure)
```

```python
plt.subplot(2, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Orijinal")

plt.subplot(2, 3, 2)
plt.imshow(prewitt, cmap="gray")
plt.title("Prewitt")

plt.subplot(2, 3, 3)
plt.imshow(sobel, cmap="gray")
plt.title("Sobel")

plt.subplot(2, 3, 4)
plt.imshow(lap, cmap="gray")
plt.title("Laplacian")

plt.subplot(2, 3, 5)
plt.imshow(log_sonuc, cmap="gray")
plt.title("LoG")

plt.subplot(2, 3, 6)
plt.imshow(canny, cmap="gray")
plt.title("Canny")

plt.show()
```
