```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/image4.jpeg')

if image is None:
    raise FileNotFoundError("image4.jpeg dosyası bulunamadı!")

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

print("Görüntü başarıyla yüklendi!")
print("Görüntü boyutu:", image.shape)

plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.title("Orijinal Görüntü")
plt.axis("off")
plt.show()
```

```python
image_075 = image.astype(np.float32) * 0.75

image_075 = np.clip(image_075, 0, 255).astype(np.uint8)

image_075_rgb = cv2.cvtColor(image_075, cv2.COLOR_BGR2RGB)

print("Görüntü piksel değerleri 0.75 ile çarpıldı!")

plt.figure(figsize=(8, 6))
plt.imshow(image_075_rgb)
plt.title("0.75 ile Çarpılmış Görüntü")
plt.axis("off")
plt.show()
```

```python
image_075_20 = image.astype(np.float32) * 0.75 + 20

image_075_20 = np.clip(image_075_20, 0, 255).astype(np.uint8)

image_075_20_rgb = cv2.cvtColor(image_075_20, cv2.COLOR_BGR2RGB)

print("Görüntü 0.75 ile çarpıldı ve 20 eklendi!")

plt.figure(figsize=(8, 6))
plt.imshow(image_075_20_rgb)
plt.title("0.75 ile Çarpılmış ve 20 Eklenmiş Görüntü")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(16, 6))

plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title("Orijinal Görüntü")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(image_075_rgb)
plt.title("0.75 ile Çarpılmış")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(image_075_20_rgb)
plt.title("0.75 ile Çarpılmış + 20")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
original_mean = np.mean(image)
mean_075 = np.mean(image_075)
mean_075_20 = np.mean(image_075_20)

print("========== SAYISAL KARŞILAŞTIRMA ==========")

print(f"Orijinal Görüntü Ortalama Piksel Değeri: {original_mean:.2f}")

print(f"0.75 ile Çarpılmış Ortalama: {mean_075:.2f}")

print(f"0.75 ile Çarpılıp 20 Eklenmiş Ortalama: {mean_075_20:.2f}")
```

```python
labels = [
    "Orijinal",
    "0.75 ile Çarpılmış",
    "0.75 ile Çarpılmış + 20"
]

values = [
    original_mean,
    mean_075,
    mean_075_20
]

plt.figure(figsize=(10, 6))

bars = plt.bar(
    labels,
    values,
    color=['blue', 'orange', 'green']
)

plt.title("Ortalama Piksel Değeri Karşılaştırması")
plt.ylabel("Ortalama Piksel Değeri")
plt.ylim(0, 255)

for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 3,
        f"{value:.2f}",
        ha='center'
    )

plt.show()
```

```python
print("========== ÖDEV SONUÇLARI ==========")

print("Görüntü: image4.jpeg")

print("\n1. İşlem:")
print("Piksel değerleri 0.75 ile çarpılmıştır.")

print("\n2. İşlem:")
print("Piksel değerleri 0.75 ile çarpılmış ve 20 eklenmiştir.")

print("\nOrijinal ortalama:", round(original_mean, 2))
print("0.75 ile çarpılmış ortalama:", round(mean_075, 2))
print("0.75 ile çarpılıp 20 eklenmiş ortalama:", round(mean_075_20, 2))

print("\nİşlemler başarıyla tamamlandı!")
```
