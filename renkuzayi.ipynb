```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image3.jpeg")
b, g, r = cv2.split(img)
```

```python
def my_bgr2rgb(b, g, r):
    return cv2.merge([r, g, b])

def my_bgr2gray(b, g, r):
    b = b.astype(np.int32)
    g = g.astype(np.int32)
    r = r.astype(np.int32)
    gri = (r * 4899 + g * 9617 + b * 1868 + 8192) >> 14
    return gri.astype(np.uint8)
```

```python
rgb_opencv = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
rgb_kendi = my_bgr2rgb(b, g, r)

gri_opencv = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gri_kendi = my_bgr2gray(b, g, r)
```

```python
hata_rgb = np.sum(np.abs(rgb_opencv.astype(int) - rgb_kendi.astype(int)))
hata_gri = np.sum(np.abs(gri_opencv.astype(int) - gri_kendi.astype(int)))

print("BGR -> RGB hata:", hata_rgb)
print("BGR -> GRAY hata:", hata_gri)
```

```python
plt.subplot(2, 2, 1)
plt.imshow(rgb_opencv)
plt.title("RGB OpenCV")

plt.subplot(2, 2, 2)
plt.imshow(rgb_kendi)
plt.title("RGB Kendi")

plt.subplot(2, 2, 3)
plt.imshow(gri_opencv, cmap="gray")
plt.title("Gri OpenCV")

plt.subplot(2, 2, 4)
plt.imshow(gri_kendi, cmap="gray")
plt.title("Gri Kendi")

plt.show()
```
