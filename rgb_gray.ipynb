```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image3.jpeg")
b, g, r = cv2.split(img)
```

```python
gri_opencv = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
def my_gray(b, g, r):
    gri = 0.299 * r + 0.587 * g + 0.114 * b
    gri = np.round(gri)
    return gri.astype(np.uint8)

gri_kendi = my_gray(b, g, r)
```

```python
fark = np.abs(gri_opencv.astype(int) - gri_kendi.astype(int))
hata = np.sum(fark)

print("Toplam hata:", hata)
```

```python
plt.subplot(1, 2, 1)
plt.imshow(gri_opencv, cmap="gray")
plt.title("OpenCV")

plt.subplot(1, 2, 2)
plt.imshow(gri_kendi, cmap="gray")
plt.title("Kendi Fonksiyonum")

plt.show()
```
