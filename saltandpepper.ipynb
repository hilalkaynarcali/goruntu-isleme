```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("image1.png")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.title("Salt and Pepper Gürültülü Resim")
plt.axis("off")
plt.show()
```

```python
median_opencv = cv2.medianBlur(image, 5)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Orijinal Gürültülü Resim")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(median_opencv)
plt.title("OpenCV Medyan Filtre 5x5")
plt.axis("off")

plt.show()
```

```python
def median_filter(image):

    height, width, channels = image.shape

    result = np.zeros_like(image)

    for i in range(2, height - 2):
        for j in range(2, width - 2):

            for k in range(channels):

                values = []

                for m in range(-2, 3):
                    for n in range(-2, 3):

                        values.append(image[i+m, j+n, k])

                values.sort()

                result[i, j, k] = values[12]

    return result
```

```python
median_manual = median_filter(image)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Gürültülü Resim")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(median_manual)
plt.title("Kendi Medyan Filtrem")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Orijinal Resim")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(median_opencv)
plt.title("OpenCV Medyan Filtre")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(median_manual)
plt.title("Kendi Medyan Filtrem")
plt.axis("off")

plt.show()
```
