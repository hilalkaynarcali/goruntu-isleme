```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sinus_uret(vektor, genlik, faz, periyot):

    N = len(vektor)

    # Vektör boyutu 2'nin kuvveti mi?
    if N == 0 or (N & (N - 1)) != 0:
        raise ValueError("Vektör boyutu 2'nin kuvveti olmalıdır!")

    if periyot <= 0:
        raise ValueError("Periyot pozitif olmalıdır!")

    # Sinüs değerlerini hesapla
    for n in range(N):

        vektor[n] = genlik * np.sin(
            (2 * np.pi * n / periyot) + faz
        )
```

```python
# Vektör boyutu
N = 128

# Başlangıçta sıfırlardan oluşan vektör
vektor = np.zeros(N)

# Sinüs parametreleri
genlik = 2
faz = np.pi / 4
periyot = 32

# Fonksiyonu çağır
sinus_uret(vektor, genlik, faz, periyot)

print("Vektör boyutu:", len(vektor))
print("Sinüs değerleri:")
print(vektor)
```

```python
n = np.arange(N)

plt.figure(figsize=(12, 4))

plt.stem(n, vektor)

plt.title("Sinüs Değerleriyle Doldurulmuş Vektör")
plt.xlabel("Örnek İndeksi (n)")
plt.ylabel("Genlik")

plt.grid(True)
plt.show()
```

```python
# Daha sık örneklenen yatay eksen
x = np.linspace(0, N - 1, 2000)

# Sinüs dalgası
y = genlik * np.sin(
    (2 * np.pi * x / periyot) + faz
)

plt.figure(figsize=(12, 4))

plt.plot(x, y)

plt.title("Sinüzoidal Dalga")
plt.xlabel("Örnek Konumu")
plt.ylabel("Genlik")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 5))

# Vektör
plt.subplot(1, 2, 1)
plt.stem(n, vektor)
plt.title("Sinüs Vektörü")
plt.xlabel("Örnek İndeksi")
plt.ylabel("Genlik")
plt.grid(True)

# Sinüzoidal dalga
plt.subplot(1, 2, 2)
plt.plot(x, y)
plt.title("Sinüzoidal Dalga")
plt.xlabel("Örnek Konumu")
plt.ylabel("Genlik")
plt.grid(True)

plt.tight_layout()

plt.savefig("odev31_sinus_sonuc.png", dpi=200)

plt.show()
```
