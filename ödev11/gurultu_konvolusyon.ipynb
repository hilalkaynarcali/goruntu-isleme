```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.webp", cv2.IMREAD_GRAYSCALE)

print("Resim boyutu:", img.shape)
print("Gürültülü resim:")
cv2_imshow(img)
```

```python
kernel = np.ones((3, 3), dtype=np.float32) / 9

satir, sutun = img.shape

padded = np.pad(img, ((1, 1), (1, 1)), mode="constant", constant_values=0)

filtreli = np.zeros((satir, sutun), dtype=np.float32)

for i in range(satir):
    for j in range(sutun):
        bolge = padded[i:i+3, j:j+3]
        filtreli[i, j] = np.sum(bolge * kernel)

filtreli = filtreli.astype(np.uint8)

print("Filtrelenmiş resim:")
cv2_imshow(filtreli)
```

```python
karsilastirma = np.hstack((img, filtreli))
cv2_imshow(karsilastirma)

cv2.imwrite("filtreli_image.webp", filtreli)
```
