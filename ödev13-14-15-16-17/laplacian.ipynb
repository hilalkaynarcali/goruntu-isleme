```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image4.jpeg", 0)
blur = cv2.GaussianBlur(img, (5, 5), 1)
```

```python
laplacian = np.array([[0,  1, 0],
                      [1, -4, 1],
                      [0,  1, 0]])

sonuc_lap = np.abs(cv2.filter2D(blur, cv2.CV_64F, laplacian))
```

```python
prewitt_x = np.array([[-1, 0, 1],
                      [-1, 0, 1],
                      [-1, 0, 1]])

sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

sonuc_prewitt = np.abs(cv2.filter2D(blur, cv2.CV_64F, prewitt_x))
sonuc_sobel = np.abs(cv2.filter2D(blur, cv2.CV_64F, sobel_x))
```

```python
plt.subplot(2, 2, 1)
plt.imshow(blur, cmap="gray")
plt.title("Gauss")

plt.subplot(2, 2, 2)
plt.imshow(sonuc_prewitt, cmap="gray")
plt.title("Prewitt")

plt.subplot(2, 2, 3)
plt.imshow(sonuc_sobel, cmap="gray")
plt.tight_layout()
plt.title("Sobel")

plt.subplot(2, 2, 4)
plt.imshow(sonuc_lap, cmap="gray")
plt.tight_layout()
plt.title("Laplacian")

plt.show()
```
