```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image4.jpeg", 0)
```

```python
blur = cv2.GaussianBlur(img, (5, 5), 1)
```

```python
kernel = np.array([[-1, -1, -1],
                   [ 0,  0,  0],
                   [ 1,  1,  1]])

sonuc = cv2.filter2D(blur, cv2.CV_32F, kernel)
sonuc = np.abs(sonuc)
```

```python
sonuc_norm = cv2.normalize(sonuc, None, 0, 255, cv2.NORM_MINMAX, dtype=cv2.CV_8U)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Orijinal")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(blur, cmap="gray")
plt.title("Gauss")
plt.axis("off")

plt.subplot(1, 3, 3)
# vmax parametresi kaldırıldı, normalize edilen değişken kullanıldı
plt.imshow(sonuc_norm, cmap="gray")
plt.title("Prewitt")
plt.axis("off")

plt.show()
```
