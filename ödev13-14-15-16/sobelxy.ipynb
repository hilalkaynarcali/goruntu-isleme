```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image4.jpeg", 0)
```

```python
blur = cv2.GaussianBlur(img, (5, 5), 1)
```

```python
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]])

sonuc_x = cv2.filter2D(blur, cv2.CV_64F, sobel_x)
sonuc_y = cv2.filter2D(blur, cv2.CV_64F, sobel_y)

sonuc_x = np.abs(sonuc_x)
sonuc_y = np.abs(sonuc_y)
```

```python
plt.subplot(1, 4, 1)
plt.imshow(img, cmap="gray")
plt.title("Orijinal")

plt.subplot(1, 4, 2)
plt.imshow(blur, cmap="gray")
plt.title("Gauss")

plt.subplot(1, 4, 3)
plt.imshow(sonuc_x, cmap="gray")
plt.title("Sobel X")

plt.subplot(1, 4, 4)
plt.imshow(sonuc_y, cmap="gray")
plt.title("Sobel Y")

plt.show()
```
