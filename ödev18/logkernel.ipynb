```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image4.jpeg", 0)
```

```python
log = np.array([[0,  0, -1, -1, -1,  0,  0],
                [0, -1, -3, -3, -3, -1,  0],
                [-1, -3,  0,  7,  0, -3, -1],
                [-1, -3,  7, 24,  7, -3, -1],
                [-1, -3,  0,  7,  0, -3, -1],
                [0, -1, -3, -3, -3, -1,  0],
                [0,  0, -1, -1, -1,  0,  0]])

sonuc = cv2.filter2D(img, cv2.CV_64F, log)
sonuc = np.abs(sonuc)
```

```python
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Orijinal")

plt.subplot(1, 2, 2)
plt.imshow(sonuc, cmap="gray")
plt.title("LoG 7x7")

plt.show()
```
