```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image3.jpeg")
b, g, r = cv2.split(img)
```

```python
def my_calchist(kanal):
    hist = np.zeros(256)
    for satir in range(kanal.shape[0]):
        for sutun in range(kanal.shape[1]):
            deger = kanal[satir, sutun]
            hist[deger] = hist[deger] + 1
    return hist
```

```python
hist_b = my_calchist(b)
hist_g = my_calchist(g)
hist_r = my_calchist(r)
```

```python
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Orijinal")

plt.subplot(2, 2, 2)
plt.plot(hist_r, color="red")
plt.title("Kırmızı Kanal")

plt.subplot(2, 2, 3)
plt.plot(hist_g, color="green")
plt.tight_layout()
plt.title("Yeşil Kanal")

plt.subplot(2, 2, 4)
plt.plot(hist_b, color="blue")
plt.tight_layout()
plt.title("Mavi Kanal")

plt.show()
```
